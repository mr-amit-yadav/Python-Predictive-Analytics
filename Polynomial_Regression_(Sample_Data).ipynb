```python
# Importing Libraries
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
```

```python
# Set random seed for reproducibility
np.random.seed(42)
```

```python
# Generate age with realistic distribution
age = np.random.normal(40, 15, 1338)
age = np.clip(age, 25, 64).astype(int)
```

```python
# Generate BMI with correlation to age
bmi_base = np.random.normal(30, 5, 1338)
bmi = bmi_base + (age - 40) * 0.1
bmi = np.clip(bmi, 18.5, 47)
```

```python
# Generate smoker status (correlated with higher charges)
smoker = np.random.choice(["yes", "no"], 1338, p=[0.2, 0.8])
```

```python
# Generate base charges considering age, BMI, and smoking
base_charges = 5000 + (age - 25) * 100  # Base charge increases with age
bmi_factor = np.where(bmi > 30, 1.5, 1.0)  # Higher charges for high BMI
smoker_factor = np.where(smoker == "yes", 2.5, 1.0)  # Much higher charges for smokers
charges = base_charges * bmi_factor * smoker_factor
charges += np.random.normal(0, 1000, 1338)  # Add some noise
```

```python
# Create the full dataset
data = pd.DataFrame({
    "age": age,
    "bmi": bmi,
    "children": np.random.randint(0, 5, 1338),
    "sex": np.random.choice(["female", "male"], 1338),
    "smoker": smoker,
    "region": np.random.choice(["northeast", "northwest", "southeast", "southwest"], 1338),
    "charges": charges
})
```

```python
# Fill missing values in the 'bmi' column with the mean
data['bmi'] = data['bmi'].fillna(data['bmi'].mean())
```

```python
# Encoding Categorical Variables
le = LabelEncoder()
data['sex'] = le.fit_transform(data['sex'])
data['smoker'] = le.fit_transform(data['smoker'])
region_df = pd.get_dummies(data['region'], drop_first=True)
```

```python
# Prepare the data
X_num = data[['age', 'bmi', 'children']].copy()
X_final = pd.concat([X_num, region_df, data['sex'], data['smoker']], axis=1)
y_final = data['charges']
```

```python
# Create polynomial features (degree=2)
poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X_final)
```

```python
# Split the data
X_train, X_test, y_train, y_test = train_test_split(X_poly, y_final, test_size=0.33, random_state=0)
```

```python
# Standardize the features
s_scaler = StandardScaler()
X_train = s_scaler.fit_transform(X_train.astype(np.float64))
X_test = s_scaler.transform(X_test.astype(np.float64))
```

```python
# Fit the polynomial regression model on the training data
# Initialize and fit the model
poly_lr = LinearRegression()
poly_lr.fit(X_train, y_train)
```

```python
# Make predictions on training and test sets
# Generate predictions
y_train_pred = poly_lr.predict(X_train)
y_test_pred = poly_lr.predict(X_test)
```

```python
# Print results (already provided)
print("Polynomial Regression (degree=2)")
print("Coefficients:", poly_lr.coef_)
print("Intercept:", poly_lr.intercept_)
print("poly train score %.3f, poly test score: %.3f"% (poly_lr.score(X_train, y_train), poly_lr.score(X_test, y_test)))
```
