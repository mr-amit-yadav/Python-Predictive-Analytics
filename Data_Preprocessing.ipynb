```python
# Load necessary Libraries
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
import matplotlib.pyplot as plt
```

```python
# Import the Dataset
data = pd.read_csv('insurance.csv')
```

```python
# Display the first 15 rows
data.head(15)
```

```python
# Handling Missing Values
# Check how many values are missing (NaN)
print("How many values are missing?")
print(data.isnull().sum())
```

```python
# Option 1: Drop the entire column with missing values
# Add code to drop the 'bmi' column and verify
data_option1 = data.copy()
data_option1.drop('bmi',axis=1, inplace =True)
print("\nOption 1: Drop the entire 'bmi' column")
print(data_option1.isnull().sum())
```

```python
# Option 2: Drop rows with missing values
# Add code to drop rows with missing values and verify
data_option2 = data.copy()
data_option2.dropna(inplace=True)
print("\nOption2: Drop rows with missing values")
print(data_option2.isnull().sum())
```

```python
# Option 3: Fill missing avlues with mean (Simple Imputer)
# Add code to fill missing values in the 'bmi' column using SimpleImputers
data_option3 = data.copy()
imputer = SimpleImputer(strategy="mean")
data_option3['bmi'] = imputer.fit_transform(data_option3[["bmi"]])
print("\nOption3: Fill missing values with mean (SimpleImputer)")
print(data_option3.isnull().sum())
```

# Data Preparation
Convert Categorical Data into Numbers
- Label Encoding: Two Distinct Values.<br>
(Example: In smoker column we have two distinct vaues with 'Yes' or 'No', that can be replaced by 1 or 0)
- One- hot encoding: Three or more distinct values. <br>(Example: In Colour 'Blue','Green','Red' that can be replace using 0,1,2)

```python
from sklearn.preprocessing import LabelEncoder

# Label Encoding: Encode 'Sex' and 'Smoker' Columns
# Create a label encoder instance and encode the 'Sex' column
le = LabelEncoder()
data['sex'] = le.fit_transform(data['sex'])
print("\nSklearn Label Encoding for 'sex':")
print(dict(zip(le.classes_, le.transform(le.classes_))))
print(data['sex'].head(10))
```

```python
# Create a label encoder instance and encode the 'smoker' column
data['smoker'] = le.fit_transform(data['smoker'])
print("n\nSklearn Label Encoding for 'smoker':")
print(dict(zip(le.classes_, le.transform(le.classes_))))
print(data['smoker'].head(10))
```

```python
from sklearn.preprocessing import OneHotEncoder

# One Hot Encoding: Encode the 'region' column
# Create a one hot encoder instance and encode the 'region' column
ohe = OneHotEncoder(sparse_output=False,drop='first')
region_encoded = ohe.fit_transform(data[['region']])
region_encoded_df = pd.DataFrame(region_encoded, columns=ohe.get_feature_names_out(['region']))
```

```python
# Convert the result into a DataFrame with appropriate column names
# The variable `region_encoded_df` is already created in the previous cell (2LKwvuGIJVop)
# and contains the one-hot encoded region data with appropriate column names.
# We will use that directly instead of creating a new `region_df`.

# Correct the pd.concat syntax by moving 'axis=1' outside the list of dataframes
data = pd.concat([data.reset_index(drop=True), region_encoded_df.reset_index(drop=True)], axis=1)
data.drop(columns=['region'], inplace=True)

print("\nSklearn One Hot Encoding for 'region':")
print(region_encoded_df.head(10))
```

```python
# Display the updated dataframe

print("\nFinal Dataframe after encoding")
print(data.head(15))
```

Train Data: Data used to train the prediction model.<br>
Test Data:
Data used to test the prediction model.

```python
# Combine numerical columns (age,bmi.children) with encoded columns
x_num = data[['age','bmi','children']].copy()
x_final = pd.concat([x_num, region_encoded_df,data['sex'], data['smoker']],axis=1)
```

```python
# Assign response variable ('charges') to y_final
y_final = data[['charges']].copy()
```

```python
# Split the data into train and test sets using train_test_split
x_train, x_test, y_train, y_test = train_test_split(x_final,y_final,test_size=0.33, random_state=0)
```

```python
# Display Shapes
print("\nShapes of train/test datasets:")
print("X_train:", x_train.shape)
print("X_test:", x_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

Feature Scaling <br>
Example: Count the number of spieces with drone.<br>
Normalization<br>
Standardization

```python
from sklearn.preprocessing import MinMaxScaler

# Noramlize the training and test sets using MinMaxScaler
n_scaler = MinMaxScaler()
x_train_normalized = n_scaler.fit_transform(x_train)
x_test_n_normalized = n_scaler.transform(x_test)
print("\nNormalized Traing Data: \n", x_train_normalized[:5])
print("\nNormalized Test Data: \n", x_test_n_normalized[:5])
```

```python
from sklearn.preprocessing import StandardScaler

# Standardize the traing and test sets using StandardScaler
s_scaler = StandardScaler()
x_train_standardized = s_scaler.fit_transform(x_train)
x_test_standardized = s_scaler.transform(x_test)
print("\nStandardized Traing Data: \n", x_train_standardized[:5])
print("\nStandardized Test Data: \n", x_test_standardized[:5])
```
